# Machine Learning Assignment 3: Classification


```python
# Libraries for working with data
import pandas as pd
import numpy as np

# Dataset
from sklearn.datasets import load_breast_cancer

# Splitting and preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Classification models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

# Model evaluation
from sklearn.metrics import accuracy_score
```

```python
# Load the Breast Cancer Wisconsin dataset
cancer = load_breast_cancer()

# Convert the feature data into a pandas DataFrame
df = pd.DataFrame(
    cancer.data,
    columns=cancer.feature_names
)

# Add the target column
df["target"] = cancer.target

# Display the first five rows
df.head()
```


## Dataset Inspection


```python
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
# Count malignant and benign cases
class_counts = df["target"].value_counts().sort_index()

class_distribution = pd.DataFrame({
    "Target": [0, 1],
    "Diagnosis": ["Malignant", "Benign"],
    "Count": [class_counts[0], class_counts[1]]
})

class_distribution
```

## Separating Features and Target

```python
# Independent variables: all 30 measurements
X = df.drop("target", axis=1)

# Target variable: tumour diagnosis
y = df["target"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

## Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

## Feature Scaling

```python
# Create the StandardScaler
scaler = StandardScaler()

# Learn the mean and standard deviation from the training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same transformation to the testing data
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)
```

## 2. Classification Algorithms

### 1. Logistic Regression


### 2. Decision Tree Classifier

### 3. Random Forest Classifier

### 4. Support Vector Machine

### 5. k-Nearest Neighbors


```python
# Create the five required classification models
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=5000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "Support Vector Machine": SVC(),

    "k-Nearest Neighbors": KNeighborsClassifier(
        n_neighbors=5
    )
}

print("All five classification models were created successfully.")
```

## 3. Model Training and Evaluation

```python
# Empty list for storing model results
results = []

# Empty dictionary for storing model predictions
predictions = {}

# Train and evaluate each classification model
for model_name, model in models.items():

    # Train the model
    model.fit(X_train_scaled, y_train)

    # Predict the diagnoses for the test data
    y_pred = model.predict(X_test_scaled)

    # Save the predictions
    predictions[model_name] = y_pred

    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)

    # Store the model result
    results.append({
        "Model": model_name,
        "Accuracy": accuracy
    })

    # Display the accuracy
    print(model_name)
    print("Accuracy:", round(accuracy, 4))
    print("-" * 40)
```

```python
# Convert the model results into a DataFrame
results_df = pd.DataFrame(results)

# Sort from highest to lowest accuracy
results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

# Convert accuracy into percentage form
results_df["Accuracy Percentage"] = (
    results_df["Accuracy"] * 100
).round(2)

# Round the original accuracy column
results_df["Accuracy"] = results_df["Accuracy"].round(4)

results_df
```

```python
# First row has the highest accuracy
best_model = results_df.iloc[0]

# Last row has the lowest accuracy
worst_model = results_df.iloc[-1]

print("BEST-PERFORMING MODEL")
print("Model:", best_model["Model"])
print("Accuracy:", best_model["Accuracy"])
print("Accuracy Percentage:", best_model["Accuracy Percentage"], "%")

print("\nWORST-PERFORMING MODEL")
print("Model:", worst_model["Model"])
print("Accuracy:", worst_model["Accuracy"])
print("Accuracy Percentage:", worst_model["Accuracy Percentage"], "%")
```
